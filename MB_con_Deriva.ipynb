# Movimiento Browniano con Deriva

Ejercicio: Encuentre la probabilidad de que el movimiento browniano con parámetro deriva, mu=0.6, y varianza, $\sigma^2$=0.25, tome valores entre 1 y 3 al tiempo 4.



1.   Analitico:




Solución:

$$
X_t = 0.6t + 0.5B_t
$$

### Entonces:

$$
P(1 < X_4 < 3)
$$

Sustituyendo $t = 4$:

$$
P(1 < 2.4 + 0.5B_4 < 3)
$$

Restando $2.4$ en los tres miembros:

$$
P(-1.4 < 0.5B_4 < 0.6)
$$

Dividiendo entre $0.5$:

$$
P(-2.8 < B_4 < 1.2)
$$

Como la varianza de $B_t$ es $t$, tenemos:

$$
B_4 \sim N(0,4)
$$

La probabilidad se calcula mediante la integral:

$$
P(-2.8 < B_4 < 1.2)
=
\int_{-2.8}^{1.2}
\frac{1}{\sqrt{2(4)\pi}}
e^{-\frac{x^2}{2(4)}}\,dx
$$

$$
\boxed{P(-2.8 < B_4 < 1.2) \approx 0.645}
$$



2.   Simulación



```python
import random as r
import numpy as np
import matplotlib.pyplot as plt
```

```python
def brown(T, n):
    lista = []
    dt = np.sqrt(T/n)
    W_0 = 0
    for i in range(1, n):
        dW = r.gauss(0, 1)
        W_1 = W_0 + dt * dW
        lista.append(W_1)
        W_0 = W_1
    lista.insert(0,0)
    return lista
```

```python
def deriva(mu,sigma,T,n):
  t=np.linspace(0,T,n)
  B=np.array(brown(T,n))
  x0=mu*t + sigma*B
  return x0.tolist()
```

```python
def proba(mu,sigma,T,n,N):
  conteo = 0
  for i in range(N):
    p = deriva(mu, sigma, T, n)
    valor_final = p[-1]
    if 1 <= valor_final <= 3:
      conteo += 1

  probabilidad = conteo / N
  return probabilidad
```

```python
proba(0.6,0.5,4,1000,1000)
```
